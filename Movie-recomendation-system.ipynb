```python
# remove stopwords(in , is , are , it ...etc) drom tags column 
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features=5000,
    stop_words='english'
)
```

```python
vector = tfidf.fit_transform(new_movies_df['tags']).toarray()
```

```python
vector
```

```python
vector.shape
```

```python

from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
similarity
```

```python
new_movies_df[new_movies_df['title'] == 'Jumanji'].index[0]
```

```python
def recommend(movie):
    movie_index = new_movies_df[new_movies_df['title'] == movie].index[0]
    distances = similarity[movie_index]
    movies_list= sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]

    for i in movies_list:
        print(new_movies_df.iloc[i[0]].title)
```

```python
recommend('Iron Man')
```

```python
# simelarity score :
movie_index = new_movies_df[new_movies_df['title'] == 'Iron Man'].index[0]

for idx, score in sorted(
    list(enumerate(similarity[movie_index])),
    key=lambda x: x[1],
    reverse=True
)[:10]:
    print(score, new_movies_df.iloc[idx].title)
```
